## 📖 Executive Summary

Welcome to the **Urban Flow Analytics** project repository! This project was developed to transform large-scale, raw taxi trip data into actionable insights, predictive models, and operational decision-support systems.

Urban transportation systems generate massive volumes of trip-level data. However, raw data often contains anomalies, missing values, and complex spatial-temporal patterns. Our solution addresses these challenges through a robust, memory-efficient, chunk-based data processing pipeline, advanced machine learning models for predictive analytics, and spatial-temporal demand forecasting.

**Key Objectives:**

* Ensure data quality and handle anomalies (negative fares, zero-distance trips, etc.).
* Develop predictive models for **Fare Estimation** and **Trip Duration**.
* Forecast hourly pickup demand across 261 taxi zones for the next 24–72 hours.
* Provide actionable business recommendations for fleet dispatching and revenue optimization.

---



## 🏗️ Project Architecture & Workflow

To handle the massive dataset (over 12 million records) without exceeding memory limits, we engineered a modular, chunk-based processing pipeline. The project is divided into distinct phases, each handled by a dedicated Jupyter Notebook.

Below is the routing guide to the core notebooks of this project. Please follow the sequence below to understand the complete data journey from raw CSVs to business intelligence.



## 🛠️ Phase 1: Preprocessing & Feature Engineering

**📂 Notebook:** notebooks/[`preprocessing_pipeline.ipynb`](https://www.google.com/search?q=./preprocessing_pipeline.ipynb)

**Purpose:**
This notebook is the foundation of our data pipeline. It handles the ingestion of massive monthly CSV files using a memory-efficient chunk-based approach.

**Key Tasks:**

* **Data Quality Assessment:** Filtering out negative fares, zero-distance trips with non-zero fares, and invalid chronologies.
* **Feature Engineering:** Creating binary flags (e.g., `is_airport_trip`), extracting temporal features, and encoding categorical variables.
* **Normalization:** Applying `RobustScaler` and `StandardScaler` incrementally to prevent data leakage and manage RAM constraints.
* **Data Splitting:** Implementing a rolling 70:30 temporal train-test split to preserve time-series integrity.



---

## 💰 Phase 2: Fare Amount Prediction

**📂 Notebook:** notebooks/[`fare_prediction_model.ipynb`](https://www.google.com/search?q=./fare_prediction_model.ipynb)

**Purpose:**
This notebook focuses on predicting the **Base Fare** of a taxi trip based on operational, temporal, and spatial features.

**Key Tasks:**

* **EDA:** Analyzing correlations between distance, duration, and fare components.
* **Modeling:** Initially testing Linear Regression, then pivoting to **Random Forest Regressor** to capture non-linear spatial and temporal pricing dynamics.
* **Optimization:** Utilizing chunk-based data loading and downcasting data types to handle the 12.2M+ records efficiently.


---

## ⏱️ Phase 3: Trip Duration Prediction (On-Time Arrival Predictor)

**📂 Notebook:** notebooks/[`On_Time_Arrival_Predictor (1).ipynb`](https://www.google.com/search?q=./On_Time_Arrival_Predictor%2520(1).ipynb)

**Purpose:**
Accurately predicting the `travel_duration` (in seconds) is crucial for rider estimates and driver scheduling. This notebook avoids target leakage by strictly excluding post-trip financial metrics.

**Key Tasks:**

* **Target Definition:** Calculating duration from `dropoff_timestamp` and `pickup_timestamp`.
* **Feature Selection:** Retaining spatial IDs, distance, and temporal features while dropping fare-related variables.
* **Modeling:** Deploying an **XGBoost Regressor** with `tree_method='hist'` for GPU-accelerated, memory-efficient training.
* **Results:** Achieved an R² of ~0.82, RMSE of ~647 seconds, and MAE of ~203 seconds.

---

## 📈 Phase 4: Demand Forecasting (UrbanFlow Demand Intelligence)

**📂 Notebook:** notebooks/[`model_training.ipynb`](https://www.google.com/search?q=./model_training.ipynb)

**Purpose:**
This notebook addresses the core operational challenge: forecasting hourly pickup demand across 261 taxi zones for a 72-hour horizon to support proactive fleet dispatching.

**Key Tasks:**

* **Spatial-Temporal Aggregation:** Aggregating 11.8M trip records into 353,963 zone-hour observations (`pickup_count`).
* **Feature Formulation:** Using deterministic temporal features (`hour_of_day`, `day_of_week`, `is_weekend`) and spatial IDs (`origin_loc_id`).
* **Modeling:** Training an **XGBoost Regression** model optimized for multi-step forecasting without recursive lag variables.
* **Results:** Achieved an exceptional R² of **0.9575**, RMSE of 14.90, and MAE of 6.46 pickups/hour.

## 📊 Phase 5: Business Dashboard & Strategic Recommendations

*(Note: Dashboard implementation is done via Tableau, connected to the forecasting outputs generated in the notebooks above).*

### 🚀 Business Recommendations

1. **Proactive Fleet Dispatch:** Use the 72-hour zone-level forecasts to reposition vehicles before demand peaks, reducing idle cruising and passenger wait times.
2. **Revenue Optimization:** Identify high-demand zones and periods to maximize trip completions and inform future dynamic pricing strategies.
3. **Urban Mobility Planning:** Utilize Origin-Destination (OD) flow insights to collaborate with city planners for designated pickup zones and traffic management.
4. **Interactive Decision Support:** The Tableau dashboard allows fleet managers to visualize actual vs. predicted demand, peak warnings, and Top-10 hotspot rankings interactively.
