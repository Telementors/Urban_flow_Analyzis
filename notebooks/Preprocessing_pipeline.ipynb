```python
import os
import pandas as pd
from IPython.display import display

folder_path = '/content/drive/MyDrive/Telementors/DATATHON/Urban_Flow_Analytics_Dataset_csv'
csv_files = sorted([f for f in os.listdir(folder_path) if f.endswith('.csv')])

all_summaries = []

for file_name in csv_files:
    file_path = os.path.join(folder_path, file_name)
    print(f"Processing: {file_name}...")

    # Read CSV
    df = pd.read_csv(file_path)
    total_rows = len(df)

    # Calculate nulls per column
    null_counts = df.isnull().sum()
    null_pct = (null_counts / total_rows) * 100

    # Summary for current file
    file_summary = pd.DataFrame({
        'Column': df.columns,
        'Null Count': null_counts.values,
        'Null %': null_pct.values.round(2)
    })

    print(f"--- {file_name} (Total Rows: {total_rows:,}) ---")
    display(file_summary)
    print("\n" + "="*50 + "\n")

    # Store data for total dataset summary
    for col, count in null_counts.items():
        all_summaries.append({
            'File': file_name,
            'Column': col,
            'Null Count': count,
            'Total Rows': total_rows
        })

# --- OVERALL SUMMARY ACROSS ALL CSV FILES ---
summary_df = pd.DataFrame(all_summaries)
overall_summary = summary_df.groupby('Column').agg(
    Total_Null_Count=('Null Count', 'sum'),
    Total_Rows=('Total Rows', 'sum')
).reset_index()

overall_summary['Overall Null %'] = (overall_summary['Total_Null_Count'] / overall_summary['Total_Rows'] * 100).round(2)

print("=== OVERALL NULL SUMMARY ACROSS ALL CSV FILES ===")
display(overall_summary)
```

```python
import os
import pandas as pd

folder_path = '/content/drive/MyDrive/Telementors/DATATHON/Urban_Flow_Analytics_Dataset_csv'
train_output = '/content/drive/MyDrive/Telementors/DATATHON/Train/train_rolling.csv'
test_output = '/content/drive/MyDrive/Telementors/DATATHON/Test/test_rolling.csv'

csv_files = sorted([f for f in os.listdir(folder_path) if f.endswith('.csv')])

first_train = True
first_test = True

# Group files in sets of 4 (3 for Train, 1 for Test)
for i in range(0, len(csv_files), 4):
    group = csv_files[i : i + 4]

    # First 3 (or available) in the group go to train
    train_group = group[:3]
    # 4th month goes to test
    test_group = group[3:] if len(group) > 3 else []

    print(f"Batch {i//4 + 1}:")
    print(f"  Train: {train_group}")
    print(f"  Test : {test_group}\n")

    # Append Train files
    for f in train_group:
        file_path = os.path.join(folder_path, f)
        for chunk in pd.read_csv(file_path, chunksize=100000):
            chunk.to_csv(train_output, mode='a', index=False, header=first_train)
            first_train = False

    # Append Test files
    for f in test_group:
        file_path = os.path.join(folder_path, f)
        for chunk in pd.read_csv(file_path, chunksize=100000):
            chunk.to_csv(test_output, mode='a', index=False, header=first_test)
            first_test = False

print("Done! Files created according to 3-train / 1-test split.")
```

```python
import pandas as pd
import os

train_path = '/content/drive/MyDrive/Telementors/DATATHON/Train/train_rolling.csv'
test_path = '/content/drive/MyDrive/Telementors/DATATHON/Test/test_rolling.csv'

def add_is_airport_trip(file_path):
    if not os.path.exists(file_path):
        print(f"File not found: {file_path}")
        return

    temp_path = file_path.replace('.csv', '_updated(1).csv')
    first_chunk = True

    print(f"Processing {file_path}...")

    # Stream in chunks to prevent memory crash
    for chunk in pd.read_csv(file_path, chunksize=100000):
        # Fill NaN values with 0 and apply logic: 1 if Airport_fee > 0 else 0
        chunk['is_airport_trip'] = (chunk['Airport_fee'].fillna(0) > 0).astype(int)

        # Save chunk to temp file
        chunk.to_csv(temp_path, mode='a', index=False, header=first_chunk)
        first_chunk = False

    # Replace the original file with the updated file
    os.replace(temp_path, file_path)
    print(f"Successfully added 'is_airport_trip' to {file_path}")

# Run feature engineering for both sets
add_is_airport_trip(train_path)
add_is_airport_trip(test_path)
```

```python
import pandas as pd

zone_path = '/content/drive/MyDrive/Telementors/DATATHON/Urban_Flow_Analytics_Zone_Dataset.csv'
zone_df = pd.read_csv(zone_path)

# 1. Fill missing values
zone_df['borough_name'] = zone_df['borough_name'].fillna('Unknown')
zone_df['service_zone'] = zone_df['service_zone'].fillna('Unknown')
zone_df['zone_name'] = zone_df['zone_name'].fillna('Unknown')

# 2. One-Hot Encoding for Low Cardinality Categoricals
zone_df = pd.get_dummies(zone_df, columns=['borough_name', 'service_zone'], prefix=['borough', 'service'], dtype=int)

# 3. Frequency Encoding for High Cardinality zone_name
zone_freq = zone_df['zone_name'].value_counts() / len(zone_df)
zone_df['zone_name_freq_enc'] = zone_df['zone_name'].map(zone_freq)

# Drop raw text zone_name if no longer needed
zone_df.drop(columns=['zone_name'], inplace=True)

print("Encoded Zone Dataset Preview:")
display(zone_df.head())
```

```python
import os
import pandas as pd

def encode_offline_flag(file_path):
    if not os.path.exists(file_path):
        print(f"File not found: {file_path}")
        return

    output_path = file_path.replace('.csv', '_encoded.csv')
    first_chunk = True

    print(f"Encoding 'offline_record_flag' in {file_path}...")

    # Map values: 'Y' -> 1, 'N' -> 0, missing/NaN -> 0
    flag_map = {'N': 0, 'Y': 1, 'n': 0, 'y': 1}

    for chunk in pd.read_csv(file_path, chunksize=100000):
        if 'offline_record_flag' in chunk.columns:
            # Map values and replace NaN with 0
            chunk['offline_record_flag'] = (
                chunk['offline_record_flag']
                .map(flag_map)
                .fillna(0)
                .astype(int)
            )

        # Save to encoded output file
        chunk.to_csv(output_path, mode='a', index=False, header=first_chunk)
        first_chunk = False

    # Overwrite the file or keep as _encoded version
    os.replace(output_path, file_path)
    print(f"Successfully encoded 'offline_record_flag' in {file_path}\n")

# Apply to train and test sets
encode_offline_flag('/content/drive/MyDrive/Telementors/DATATHON/Train/train_rolling.csv')
encode_offline_flag('/content/drive/MyDrive/Telementors/DATATHON/Test/test_rolling.csv')
```

```python
import os
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# ---------------------------------------------------------
# 1. FILE PATHS & SETUP
# ---------------------------------------------------------
train_path = '/content/drive/MyDrive/Telementors/DATATHON/Train/train_rolling.csv'
test_path = '/content/drive/MyDrive/Telementors/DATATHON/Test/test_rolling.csv'

train_output = '/content/drive/MyDrive/Telementors/DATATHON/Train/train_final.csv'
test_output = '/content/drive/MyDrive/Telementors/DATATHON/Test/test_final.csv'

# Columns to drop due to extreme multicollinearity
cols_to_drop = ['charge_total', 'is_airport_trip']

# Numerical columns to scale
scale_cols = ['distance_miles', 'base_fare', 'driver_tip_payment', 'toll_total', 'surcharge_misc']

# Mapping for offline_record_flag
flag_map = {'N': 0, 'Y': 1, 'n': 0, 'y': 1}

# ---------------------------------------------------------
# 2. STEP 1: FIT SCALER ON TRAINING DATA ONLY (CHUNK-WISE)
# ---------------------------------------------------------
print("1. Fitting StandardScaler on Training set only...")
scaler = StandardScaler()

# First pass on Train: Incremental fitting using partial_fit to save RAM
for chunk in pd.read_csv(train_path, chunksize=100000):
    # Filter invalid records before fitting
    chunk = chunk[(chunk['base_fare'] > 0) & (chunk['distance_miles'] > 0)]

    # Partial fit on numerical features
    if not chunk.empty:
        scaler.partial_fit(chunk[scale_cols])

print("StandardScaler fitting complete.")

# ---------------------------------------------------------
# 3. STEP 2: PROCESS & TRANSFORM TRAINING SET
# ---------------------------------------------------------
print("\n2. Processing and Transforming Training set...")
first_chunk = True

for chunk in pd.read_csv(train_path, chunksize=100000):
    # a. Filter outliers / invalid rows
    chunk = chunk[(chunk['base_fare'] > 0) & (chunk['distance_miles'] > 0)].copy()
    if chunk.empty:
        continue

    # b. Encode offline_record_flag
    if 'offline_record_flag' in chunk.columns:
        chunk['offline_record_flag'] = chunk['offline_record_flag'].map(flag_map).fillna(0).astype(int)

    # c. Drop multicollinear columns
    chunk.drop(columns=[col for col in cols_to_drop if col in chunk.columns], inplace=True)

    # d. Apply Scaling (fitted on train)
    chunk[scale_cols] = scaler.transform(chunk[scale_cols])

    # e. Save to final train file
    chunk.to_csv(train_output, mode='a', index=False, header=first_chunk)
    first_chunk = False

print(f"Training set processed and saved to: {train_output}")

# ---------------------------------------------------------
# 4. STEP 3: PROCESS & TRANSFORM TEST SET (USING TRAIN SCALER)
# ---------------------------------------------------------
print("\n3. Processing and Transforming Test set...")
first_chunk = True

for chunk in pd.read_csv(test_path, chunksize=100000):
    # a. Filter outliers / invalid rows
    chunk = chunk[(chunk['base_fare'] > 0) & (chunk['distance_miles'] > 0)].copy()
    if chunk.empty:
        continue

    # b. Encode offline_record_flag
    if 'offline_record_flag' in chunk.columns:
        chunk['offline_record_flag'] = chunk['offline_record_flag'].map(flag_map).fillna(0).astype(int)

    # c. Drop multicollinear columns
    chunk.drop(columns=[col for col in cols_to_drop if col in chunk.columns], inplace=True)

    # d. Apply Scaling (TRANSFORM ONLY using train scaler)
    chunk[scale_cols] = scaler.transform(chunk[scale_cols])

    # e. Save to final test file
    chunk.to_csv(test_output, mode='a', index=False, header=first_chunk)
    first_chunk = False

print(f"Test set processed and saved to: {test_output}")
print("\nPipeline execution completed successfully!")
```

```python
import os
import pandas as pd
import numpy as np

train_path = '/content/drive/MyDrive/Telementors/DATATHON/Train/train_final.csv'
test_path = '/content/drive/MyDrive/Telementors/DATATHON/Test/test_final.csv'

train_clean_path = '/content/drive/MyDrive/Telementors/DATATHON/Train/train_cleaned.csv'
test_clean_path = '/content/drive/MyDrive/Telementors/DATATHON/Test/test_cleaned.csv'

# Features to drop:
# 1. 'base_fare': Highly redundant with distance_miles (corr = 0.94)
# 2. Constant/zero-variance features like 'offline_record_flag' if they carry no signal
cols_to_drop = ['base_fare']

def clean_and_remove_redundancy(file_in, file_out):
    if not os.path.exists(file_in):
        print(f"File not found: {file_in}")
        return

    first_chunk = True
    print(f"Processing {file_in}...")

    for chunk in pd.read_csv(file_in, chunksize=100000):
        # 1. Drop high multicollinearity columns
        chunk.drop(columns=[c for c in cols_to_drop if c in chunk.columns], inplace=True)

        # 2. Check for constant columns (std == 0) and drop them dynamically if needed
        # (e.g., if offline_record_flag is all 0s)
        for col in chunk.columns:
            if chunk[col].nunique() <= 1 and col in ['offline_record_flag']:
                chunk.drop(columns=[col], inplace=True)

        # Save cleaned chunk
        chunk.to_csv(file_out, mode='a', index=False, header=first_chunk)
        first_chunk = False

    print(f"Saved cleaned dataset to: {file_out}\n")

# Run cleanup on both train and test
clean_and_remove_redundancy(train_path, train_clean_path)
clean_and_remove_redundancy(test_path, test_clean_path)
```
